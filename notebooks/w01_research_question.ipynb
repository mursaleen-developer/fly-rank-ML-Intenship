## 1. My lane (or freestyle) and why

**Lane: Refresh / Content Opportunity Scoring.**

I will investigate whether the available content and performance signals can help identify pages that may be worth reviewing or refreshing first. This lane fits the starter data because it contains page-level content characteristics, recent and previous performance, content age, freshness, CTR, average position, and trend information. The goal is not to automatically decide that a page should be changed, but to create useful decision support for prioritizing which pages a content team should review.

```python
import pandas as pd

DATA_URL = "https://raw.githubusercontent.com/mursaleen-developer/fly-rank-ML-Intenship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(DATA_URL)
print(f"Starter dataset shape: {df.shape[0]:,} rows x {df.shape[1]} columns")
```

## 2. The question: decision, action, cost of a wrong call

**Research question:** Can content and historical performance signals help prioritize which pages should be reviewed for a possible refresh?

**Decision:** A content or SEO team needs to decide which pages should receive attention first.

**Action:** The team could review the highest-priority pages, investigate the reason signals, and decide whether to refresh, leave unchanged, or gather more information. The output is intended to support human review rather than automatically changing content.

**Cost of a wrong call:** If a page is incorrectly prioritized, the team may spend limited time and effort refreshing a page that did not need attention. If a page that could benefit from review is missed, an opportunity to improve its performance may be delayed.

```python
# Check that the fields needed for this question are available.
required = ["content_id", "impressions_90d", "impressions_last_30d", "impressions_prev_30d", "trend_direction", "trend_pct"]
missing = [c for c in required if c not in df.columns]
print("Required fields available:", not missing)
if missing:
    print("Missing:", missing)
```

## 3. Quick look at the data (2-3 real numbers)

The starter dataset contains **30,000 content pages**. In the current snapshot, **16,262 pages (54.2%) have a downward trend**. Among pages with at least 1,000 impressions in the 90-day period, **8,031 pages are trending down**, which is **59.4% of those 13,512 higher-exposure pages**. These numbers make the refresh-opportunity lane worth investigating because there is a substantial set of pages with meaningful exposure that are currently showing declining performance.

These are descriptive observations from the starter dataset, not evidence that refreshing a page will cause its performance to improve.

```python
# Load/check the starter data and calculate the supporting numbers used above.
total_pages = len(df)
down_pages = (df["trend_direction"] == "down").sum()
down_pct = down_pages / total_pages * 100
higher_exposure = df["impressions_90d"] >= 1000
higher_exposure_pages = higher_exposure.sum()
down_higher_exposure = ((df["trend_direction"] == "down") & higher_exposure).sum()
down_higher_exposure_pct = down_higher_exposure / higher_exposure_pages * 100

print(f"Total pages: {total_pages:,}")
print(f"Pages with downward trend: {down_pages:,} ({down_pct:.1f}%)")
print(f"Pages with >=1,000 impressions_90d: {higher_exposure_pages:,}")
print(f"Downward-trend pages among them: {down_higher_exposure:,} ({down_higher_exposure_pct:.1f}%)")
```
