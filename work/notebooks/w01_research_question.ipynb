## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

DATA_URL = "https://raw.githubusercontent.com/mursaleen-developer/fly-rank-ML-Intenship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(DATA_URL)

total_pages = len(df)
down_pages = (df["trend_direction"] == "down").sum()
down_pct = down_pages / total_pages * 100

higher_exposure = df["impressions_90d"] >= 1000
higher_exposure_pages = higher_exposure.sum()

down_higher_exposure = (
    (df["trend_direction"] == "down") & higher_exposure
).sum()

print(f"Total pages: {total_pages:,}")
print(f"Pages with downward trend: {down_pages:,} ({down_pct:.1f}%)")
print(f"Pages with >=1,000 impressions: {higher_exposure_pages:,}")
print(f"Downward-trend pages among them: {down_higher_exposure:,}")
```
